```python
# Take input from user
salary = float(input("Enter your salary: "))

# Apply conditions
if salary < 30000:
    tax_rate = 0.05
elif salary <= 70000:
    tax_rate = 0.15
else:
    tax_rate = 0.25

# Calculate tax
tax = salary * tax_rate

# Display result
print("Salary:", salary)
print("Tax Rate:", tax_rate * 100, "%")
print("Final Tax:", tax)
```

```python
class User:
    def __init__(self, name):
        self.name = name

class Message:
    def __init__(self, sender, text):
        self.sender = sender
        self.text = text

class ChatRoom:
    def __init__(self):
        self.users = []
        self.messages = []

    def join(self, user):
        self.users.append(user)
        print(user.name, "joined the chat")

    def leave(self, user):
        self.users.remove(user)
        print(user.name, "left the chat")

    def send_message(self, user, text):
        message = Message(user, text)
        self.messages.append(message)

    def show_chat(self):
        for msg in self.messages:
            print(msg.sender.name + ":", msg.text)


# Example Usage
u1 = User("Ali")
u2 = User("Ahmed")

chat = ChatRoom()

chat.join(u1)
chat.join(u2)

chat.send_message(u1, "Hello!")
chat.send_message(u2, "Hi!")

chat.show_chat()

chat.leave(u1)
```
