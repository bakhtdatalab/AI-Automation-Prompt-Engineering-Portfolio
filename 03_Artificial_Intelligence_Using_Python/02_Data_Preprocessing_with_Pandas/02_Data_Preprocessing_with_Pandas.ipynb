```python
# 1-Dataset Loading

# I. Load the dataset into a Pandas DataFrame
import pandas as pd
df = pd.read_csv("HousePricePrediction.csv")

# II. •	Display the first 10 rows
print("First 10 rows:")
print(df.head(10))
```

```python
# 2-Data Exporation

# I. Display the shape of the dataset
print("\nShape of dataset:")
print(df.shape)

# II. Check data types of all columns
print("\nData Types:")
print(df.dtypes)

# III. Generate summary statistics of numerical features
print("\nSummary Statistics:")
print(df.describe())
```

```python
# 3-Data Cleaning

# I. Identify/check missing values
print("\nMissing Values:\n", df.isnull().sum())

# II. Handle missing values appropriately

# Drop records with null values
df = df.dropna()  # you can also use fillna()

# Replacing SalePrice empty values with their mean values
df['SalePrice'] = df['SalePrice'].fillna(
  df['SalePrice'].mean())

# III. Check and remove duplicate records if any
df = df.drop_duplicates()

df.isnull().sum()
df.head()
```

```python
# 4-Feature Selection

# I. # Drop unnecessary columns (example: Id)
# Drop Id as it doesn't contribute in price
df.drop(['Id'],
             axis=1,
             inplace=True)
```

```python
# 5-Data Preprocessing

# I. Convert categorical variables into numerical form (encoding)_(Covered in Supervised ML Part2)
cols = ['MSZoning', 'LotConfig', 'BldgType', 'Exterior1st']
df = pd.get_dummies(df, columns=cols, drop_first=True)

# II. Final dataset prepared for Machine Learning, preview:
print("\nProcessed Data:\n", df.head())
```
