The AI creates product marketing content using:

## Project Goal

The goal of this project is to automate the product-copywriting process by converting structured product information into clear, accurate, audience-focused, and e-commerce-ready product descriptions.


## Step 2.1: Create Product Inputs and the AI Prompt

## Prompt Design and Core Logic

In this section, we define the product and marketing information required to generate an accurate, engaging, and e-commerce-ready product description.


```python
# Product and marketing information

product_name = "EcoSip Reusable Water Bottle"

product_category = "Sustainable drinkware"

product_features = """
- Made from stainless steel
- Double-wall insulated design
- Leak-resistant lid
- 750 ml capacity
- Reusable and designed for everyday use
"""

target_audience = """
Environmentally conscious students and working professionals
who want a practical reusable water bottle
"""

brand_tone = "Friendly, modern, and trustworthy"

selling_goal = """
Encourage customers to purchase the product from the online store
"""
```

```python
# Create a structured prompt for AI product-description generation

prompt = f"""
You are an expert e-commerce copywriter, product marketer,
and SEO content specialist.

Your task is to create an accurate, engaging, audience-focused,
and conversion-oriented product description using only the
product information provided below.

Product Name:
{product_name}

Product Category:
{product_category}

Product Features:
{product_features}

Target Audience:
{target_audience}

Brand Tone:
{brand_tone}

Primary Selling Goal:
{selling_goal}

Follow these instructions carefully:

1. Use only the product information provided.
2. Do not invent specifications, materials, certifications,
   warranties, discounts, customer reviews, or product results.
3. Explain how the listed features may be useful to the target audience.
4. Use the requested brand tone consistently.
5. Write clear, natural, and customer-friendly language.
6. Avoid exaggerated or unsupported claims.
7. Focus on relevant benefits without changing the product facts.
8. Make the content suitable for an e-commerce product page.
9. Include a clear but natural call to action.
10. Suggest relevant SEO keywords based only on the provided information.

Use exactly this output structure:

## Product Title

## Short Description

## Detailed Product Description

## Key Features

## Customer Benefits

## Call to Action

## Suggested SEO Keywords
"""
```

```python
print(prompt)
```

## Step 2.2: AI-Powered Product Description Generation

In this section, the structured product prompt is sent to the OpenAI API. The AI analyzes the product information and generates an accurate, engaging, audience-focused, and e-commerce-ready product description.


```python
!pip install -q openai
```

```python
from google.colab import userdata

api_key = userdata.get("OPENAI_API_KEY")
```

```python
from openai import OpenAI

client = OpenAI(api_key=api_key)
```

```python
# gerate the product description
response = client.responses.create(
    model="gpt-5-mini",
    input=prompt
)

product_description = response.output_text

print(product_description)
```

## Refine the Prompt

```python
# Refined prompt for AI product-description generation

prompt = f"""
You are an expert e-commerce copywriter, product marketer,
SEO content specialist, and customer-focused product writer.

Your task is to create an accurate, clear, engaging, and
conversion-oriented product description using only the
product information provided below.

PRODUCT INFORMATION

Product Name:
{product_name}

Product Category:
{product_category}

Product Features:
{product_features}

TARGET CUSTOMER

Target Audience:
{target_audience}

BRAND REQUIREMENTS

Brand Tone:
{brand_tone}

Primary Selling Goal:
{selling_goal}

Follow these rules carefully:

1. Use only the product facts provided.
2. Do not invent specifications, dimensions, materials,
   certifications, warranties, discounts, prices,
   customer reviews, ratings, or product results.
3. Do not claim that the product is the "best," "perfect,"
   "guaranteed," or superior to competitors.
4. Clearly distinguish between product features and
   customer benefits.
5. Explain benefits only when they are reasonably supported
   by the listed features.
6. Write for the specified target audience.
7. Use the requested brand tone consistently.
8. Keep the language clear, natural, and easy to scan.
9. Avoid unnecessary repetition and generic marketing phrases.
10. Make the content suitable for an e-commerce product page.
11. Use concise paragraphs and readable bullet points.
12. Include a clear, natural, and non-misleading call to action.
13. Suggest SEO keywords based only on the provided product
    name, category, features, and intended use.
14. Do not add facts that were not provided.

Use exactly this output structure:

## Product Title

Create a clear and descriptive product title.
Do not add unsupported features.

## Short Description

Write 2 to 3 concise sentences.

## Detailed Product Description

Write 1 to 2 clear paragraphs that explain the product,
its main features, and its relevant uses.

## Key Features

List the provided product features as clear bullet points.

## Customer Benefits

List practical benefits supported by the provided features.

## Call to Action

Write one concise and natural call to action.

## Suggested SEO Keywords

Provide 8 to 12 relevant keywords or keyword phrases.
"""
```


## Interactive Product Description Generator

## Generate the Interactive Product Description

```python
response = client.responses.create(
    model="gpt-5-mini",
    input=prompt
)

product_description = response.output_text

print(product_description)
```
