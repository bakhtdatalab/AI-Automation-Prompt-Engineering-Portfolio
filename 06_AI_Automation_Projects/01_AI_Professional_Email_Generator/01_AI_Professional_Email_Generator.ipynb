# AI-Powered Professional Email Generator

## Project 1 — AI Automation Portfolio

This project uses Python and prompt engineering to generate professional emails based on user requirements.

## Step 2: Prompt Design

In this section, we define the information required to generate a professional email and create a structured prompt for the AI model.

```python
# Email information provided by the user

email_purpose = "Request a meeting"
recipient = "Hiring Manager"
key_details = "I would like to discuss a potential internship opportunity next week."
tone = "Professional"
```

```python
# Create a structured prompt for the AI model

prompt = f"""
You are a professional business communication assistant.

Write a clear and professional email using the following information:

Email purpose: {email_purpose}
Recipient: {recipient}
Key details: {key_details}
Tone: {tone}

Requirements:
1. Create a relevant email subject line.
2. Include a professional greeting.
3. Write a clear and concise email body.
4. Use the requested tone.
5. End with a polite closing.
6. Do not add explanations before or after the email.

Return only the completed email.
"""
```

```python
print(prompt)
```

## Step 2.2: Connect to the OpenAI API

In this section, we connect the structured prompt to an AI model and generate a professional email.

```python
!pip install -q openai
```

```python
from google.colab import userdata

api_key = userdata.get("OPENAI_API_KEY")
```

```python
from openai import OpenAI

client = OpenAI(api_key=api_key)
```

```python
client = OpenAI(api_key=api_key)
```

```python
response = client.responses.create(
    model="gpt-5-mini",
    input=prompt
)

generated_email = response.output_text

print(generated_email)
```

## Improve the Prompt

Now we will make the prompt more specific and reliable.

We will replace the current prompt code with the following improved version:

```python
prompt = f"""
You are an expert business communication assistant.

Your task is to write a polished, professional, and ready-to-send email.

Use the following information:

Email purpose: {email_purpose}
Recipient: {recipient}
Key details: {key_details}
Requested tone: {tone}

Follow these instructions carefully:

1. Create a concise and relevant subject line.
2. Begin with an appropriate professional greeting.
3. Clearly communicate the email's purpose.
4. Include all relevant information from the key details.
5. Use the requested tone consistently.
6. Keep the email concise and easy to read.
7. Avoid unnecessary repetition, exaggerated language, and generic filler.
8. End with a polite professional closing.
9. Add "Bakht Zamin" as a signature placeholder.
10. Return only the completed email.
11. Do not explain your writing choices.
12. Do not use markdown formatting.

Output format:

Subject: [Subject line]

[Greeting]

[Email body]

[Professional closing]
[Your Name]
"""
```
