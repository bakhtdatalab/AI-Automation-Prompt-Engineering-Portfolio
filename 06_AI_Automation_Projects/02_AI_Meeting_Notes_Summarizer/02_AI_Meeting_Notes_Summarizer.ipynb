# AI Meeting Notes Summarizer

## Project 2 — AI Automation Portfolio

This project uses Python, Prompt Engineering, and the OpenAI API to convert unstructured meeting notes into clear and organized business summaries.

The AI identifies:

* Main discussion points
* Key decisions
* Action items
* Responsible people
* Deadlines
* Follow-up tasks


## Project Goal

The goal of this project is to automate the process of reviewing meeting notes and converting them into a structured summary that helps teams quickly understand what was discussed, what decisions were made, and what actions need to be completed.


## Step 2: Prompt Design and Core Logic

In this section, we create sample meeting notes and design a structured prompt that instructs the AI to identify and organize important meeting information.


```python
# Sample unstructured meeting notes

meeting_notes = """
The marketing team met on August 1 to discuss the upcoming product launch.

Sameer explained that the social media campaign is approximately 70% complete.
He will finish the remaining content and submit it for review by August 8.

Hassan reported that the advertising budget is still being reviewed.
He will share the final budget with the team by August 6.

The team agreed to launch the new product on September 15.

The website landing page needs additional improvements before the launch.
Ali will coordinate with the web development team and provide an update during
the next meeting.

The next marketing meeting is scheduled for August 10.
"""
```

```python
# Create a structured prompt for meeting summarization

prompt = f"""
You are an expert business meeting assistant.

Your task is to analyze the meeting notes below and create a clear,
accurate, and well-organized meeting summary.

Meeting Notes:
{meeting_notes}

Follow these instructions:

1. Write a short overview of the meeting.
2. Identify the main discussion points.
3. List all important decisions.
4. Extract every action item.
5. Include the responsible person for each action item when available.
6. Include deadlines when they are mentioned.
7. Identify any follow-up tasks.
8. Do not invent information that is not included in the meeting notes.
9. If information is missing, write "Not specified."
10. Use clear headings and bullet points.
11. Keep the summary concise and professional.

Use this output structure:

## Meeting Overview

## Main Discussion Points

## Key Decisions

## Action Items

## Follow-Up Items

## Next Meeting
"""
```

```python
print(prompt)
```

## Step 2.2: AI-Powered Meeting Summary

In this section, the structured meeting prompt is sent to the OpenAI API. The AI analyzes the meeting notes and generates a clear, organized, and professional summary.


```python
!pip install -q openai
```

```python
from google.colab import userdata

api_key = userdata.get("OPENAI_API_KEY")
```

```python
from openai import OpenAI

client = OpenAI(api_key=api_key)
```

```python
response = client.responses.create(
    model="gpt-5-mini",
    input=prompt
)

meeting_summary = response.output_text

print(meeting_summary)
```

# Refine the Prompt

Now we’ll improve the prompt so the AI handles action items and missing information more consistently.

Replacing the current prompt with the following refined code version:

```python
# Refined prompt for structured meeting summarization

prompt = f"""
You are an expert business meeting analyst.

Your task is to analyze the meeting notes and create an accurate,
concise, professional, and action-oriented meeting summary.

Meeting Notes:
{meeting_notes}

Follow these rules carefully:

1. Use only information provided in the meeting notes.
2. Do not invent names, dates, decisions, deadlines, or responsibilities.
3. Clearly separate completed work from future action items.
4. Extract every action item mentioned in the notes.
5. For each action item, identify:
   - Task
   - Responsible person
   - Deadline
6. If a responsible person or deadline is not mentioned, write
   "Not specified."
7. Include important project updates and completed work.
8. Include all confirmed decisions.
9. Include important follow-up items.
10. Keep the summary concise and easy to scan.
11. Use professional language.
12. Use bullet points where appropriate.

Use exactly this output structure:

## Meeting Overview

## Key Updates

## Main Discussion Points

## Key Decisions

## Action Items

| Task | Responsible Person | Deadline |
|---|---|---|

## Follow-Up Items

## Next Meeting
"""
```

## Why This Prompt Is Better

The original prompt asked the AI to identify important information.

The refined prompt adds stronger instructions for:

* Separating completed work from future tasks
* Extracting action items consistently
* Organizing responsibilities and deadlines in a table
* Avoiding invented information
* Handling missing information clearly
* Producing an action-oriented business summary
