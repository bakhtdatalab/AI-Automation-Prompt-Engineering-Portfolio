# AI Content Calendar Generator

## Project 3 — AI Automation Portfolio

This project uses Python, Prompt Engineering, and the OpenAI API to generate structured social media content calendars.

The AI creates content ideas based on:

* Business or brand information
* Industry or niche
* Target audience
* Marketing goals
* Social media platform
* Content duration


## Project Goal

The goal of this project is to automate the content-planning process by converting business and marketing requirements into a clear, organized, and actionable social media content calendar.


## Create Marketing Inputs
###  Prompt Design and Core Logic

In this section, we define the business and marketing information required to generate a structured social media content calendar.


### Adding the Business and Marketing Variables

```python
# Business and marketing information

business_name = "EcoStyle"

industry = "Sustainable fashion"

target_audience = "Environmentally conscious young adults aged 18 to 35"

marketing_goal = "Increase brand awareness and encourage website visits"

platform = "Instagram"

duration = 7
```

## Create the AI Content Calendar Prompt

```python
# Create a structured prompt for the AI content calendar

prompt = f"""
You are an expert social media strategist and digital marketing specialist.

Your task is to create a practical, engaging, and platform-specific
content calendar based on the business information below.

Business Name: {business_name}

Industry: {industry}

Target Audience: {target_audience}

Marketing Goal: {marketing_goal}

Social Media Platform: {platform}

Content Calendar Duration: {duration} days

Follow these instructions:

1. Create one content idea for each day.
2. Make every idea relevant to the business and target audience.
3. Align the content with the marketing goal.
4. Use content formats that work well on the selected platform.
5. Include a variety of content types.
6. Avoid repeating the same content idea.
7. Balance educational, engaging, promotional, and brand-building content.
8. Include a clear call to action for each post.
9. Suggest relevant hashtags.
10. Keep all ideas practical and realistic.

Use this output structure:

## Content Strategy Overview

## {duration}-Day Content Calendar

| Day | Content Topic | Content Format | Content Idea | Call to Action | Hashtags |
|---|---|---|---|---|---|
"""
```

```python
print(prompt)
```

## Generate the AI Content Calendar

### Step 2.2: AI-Powered Content Calendar Generation

In this section, the structured marketing prompt is sent to the OpenAI API. The AI analyzes the business information and generates a practical, platform-specific content calendar.


```python
!pip install -q openai
```

```python
# Loading the   API key
from google.colab import userdata

api_key = userdata.get("OPENAI_API_KEY")
```

```python
# Creating the OpenAI Client
from openai import OpenAI

client = OpenAI(api_key=api_key)
```

### Generate the Content Calendar

```python
response = client.responses.create(
    model="gpt-5-mini",
    input=prompt
)

content_calendar = response.output_text

print(content_calendar)
```

## Refine the Prompt

Now we’ll improve the prompt so the generated calendar is more strategic, practical, and consistent.

Replacing  our current prompt with:

```python
# Refined prompt for AI content calendar generation

prompt = f"""
You are an expert social media strategist, digital marketer,
and content planner.

Your task is to create a practical, strategic, engaging, and
platform-specific content calendar using the business information below.

Business Name:
{business_name}

Industry:
{industry}

Target Audience:
{target_audience}

Primary Marketing Goal:
{marketing_goal}

Social Media Platform:
{platform}

Content Calendar Duration:
{duration} days

Follow these rules carefully:

1. Create exactly one content idea for each day.
2. Make every idea relevant to the business, industry, and target audience.
3. Align every post with the primary marketing goal.
4. Use content formats that are suitable for {platform}.
5. Include a balanced mix of:
   - Educational content
   - Engaging or interactive content
   - Trust-building content
   - Brand-awareness content
   - Product or service promotion
6. Avoid repeating the same topic, format, or content angle.
7. Make the ideas practical for a small business or beginner content creator.
8. Include a clear and specific call to action for every post.
9. Suggest 3 to 5 relevant hashtags for every post.
10. Avoid generic ideas that could apply to any business.
11. Do not invent product features, discounts, customer results,
    or business facts that were not provided.
12. Keep each calendar entry concise and actionable.

First, provide a short content strategy overview that explains:

- The main content themes
- How the content supports the marketing goal
- How the content is adapted for {platform}

Then create the calendar using exactly this format:

## Content Strategy Overview

## {duration}-Day Content Calendar

| Day | Content Pillar | Topic | Format | Content Idea | Call to Action | Hashtags |
|---|---|---|---|---|---|---|
"""
```

## The refined prompt adds:

- Content pillars to make the strategy more organized
- A clearer balance between education, engagement, trust, awareness, and promotion
- Platform-specific requirements
- Practicality for small businesses
- A rule against generic ideas
- A rule against inventing business claims
- More specific hashtag requirements
- A strategic overview before the calendar
