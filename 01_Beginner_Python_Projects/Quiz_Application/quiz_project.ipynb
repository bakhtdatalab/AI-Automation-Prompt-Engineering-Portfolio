# ❓ Python Quiz Application

A simple command-line quiz application created with Python in Google Colab. The application asks multiple-choice questions, evaluates user responses, and displays the final score.

## 📌 Project Overview

The quiz application presents a series of questions to the user and checks each answer before calculating the final result.

This project was created to practice Python programming concepts related to loops, lists, dictionaries, and conditional statements.

## 🚀 Features

* Multiple-choice questions
* Automatic score calculation
* Instant answer validation
* Final score display
* Interactive command-line interface

```python
score = 0
print("Welcome! to the Quiz Application!")
if input("Are you ready to play? (yes/no): ").lower() == "yes":
  # Question 1
  answer1 = input("What is the capital of Pakistan? ")
  if answer1.lower() == "islamabad":
    score += 1
    print("Correct!")
  else:
    print("Incorrect. The correct answer is Islamabad.")

  # Question 2
  answer2 = input("What is the largest planet in our solar system? ")
  if answer2.lower() == "jupiter":
    score += 1
    print("Correct!")
  else:
    print("Incorrect. The correct answer is Jupiter.")

  # Question 3
  answer3 = input("What is the chemical symbol for gold? ")
  if answer3.lower() == "au":
    score += 1
    print("Correct!")
  else:
    print("Incorrect. The correct answer is Au.")

  # Question 4
  answer4 = input("Question 2: What is 5 + 7? ")
  if answer4 == "12":
    score += 1
    print("Correct!")
  else:
    print("Incorrect. The correct answer is 12.")

  # Question 5
  answer5 = input("What is the smallest country in the world? ")
  if answer5.lower() == "vatican city":
    score += 1
    print("Correct!")
  else:
    print("Incorrect. The correct answer is Vatican City.")

# Showing the Quiz result out of 5
print("Quiz completed! Your score is:", score, "Out of 5")
```
